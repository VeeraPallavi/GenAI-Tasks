## 🌱 What is Generative AI?

Generative AI refers to models that can **generate new content**:
- Text
- Images
- Code
- Audio

Unlike traditional ML (predict/classify), GenAI **creates**.

### Examples:
- ChatGPT → text generation
- DALL·E → image generation
- GitHub Copilot → code completion

## 🧠 Large Language Models (LLMs)

LLMs are deep learning models trained on massive text corpora.

Examples:
- BERT → Understanding-focused
- GPT → Generation-focused
- T5 → Text-to-text framework

Key properties:
- Billions of parameters
- Trained using self-supervised learning
- Use Transformer architecture

## 🔄 Transformer Architecture

Transformers replaced RNNs/LSTMs by enabling:
- Parallel processing
- Long-range dependency capture

### Core Components:
- Token Embeddings
- Positional Encoding
- Multi-Head Self-Attention
- Feed Forward Networks

## 👀 Attention Mechanism (Self-Attention)

Self-attention allows each word to **attend to all other words**.

It computes:
- Query (Q)
- Key (K)
- Value (V)

This helps the model understand **context** and **relationships**.

## 🧠 Prompt Engineering

Prompt engineering is the art of **designing effective inputs** for LLMs.

Good prompts:

- Provide context
- Specify format

## 🎯 Prompting Techniques

1. **Zero-shot Prompting**
   - No examples

2. **One-shot Prompting**
   - One example

3. **Few-shot Prompting**
   - Multiple examples

## ⚠️ Model Limitations – Hallucinations

LLMs may:
- Generate incorrect facts
- Sound confident but be wrong

Mitigation:
- Grounding with data
- Retrieval-Augmented Generation (RAG)
- Explicit constraints in prompts

```python
from groq import Groq

client = Groq(api_key=groq_api_key)

paragraph = """When Percy Jackson gets an urgent distress call from his friend Grover, he immediately prepares for battle. He knows he will need his powerful demigod allies at his side, his trusty bronze sword Riptide, and… a ride from his mom.

The demigods rush to the rescue to find that Grover has made an important discovery: two powerful half-bloods whose parentage is unknown. But that’s not all that awaits them. The titan lord Kronos has devised his most treacherous plot yet, and the young heroes have just fallen prey.

They’re not the only ones in danger. An ancient monster has arisen — one rumored to be so powerful it could destroy Olympus — and Artemis, the only goddess who might know how to track it, is missing. Now Percy and his friends, along with the Hunters of Artemis, have only a week to find the kidnapped goddess and solve the mystery of the monster she was hunting. Along the way, they must face their most dangerous challenge yet: the chilling prophecy of the titan’s curse."""
```

1️⃣ Summarizing

```python
summary_prompt = f"Summarize the following text in 5 bullet points:\n\n{paragraph}"

summary = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": summary_prompt}]
)

print("\n--- SUMMARY ---")
print(summary.choices[0].message.content)
```

2️⃣ Inferring

```python
infer_prompt = f"What can be inferred about the reader's sentiment from the following text?\n\n{paragraph}"

infer = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": infer_prompt}]
)

print("\n--- INFERENCE ---")
print(infer.choices[0].message.content)
```

3️⃣ Transforming

```python
transform_prompt = f"Convert the following paragraph into a movie script:\n\n{paragraph}"

transform = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": transform_prompt}]
)

print("\n--- MOVIE SCRIPT ---")
print(transform.choices[0].message.content)
```

4️⃣ Expanding

```python
expand_prompt = f"Expand this idea into a detailed explanation:\n\n{paragraph}"

expand = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": expand_prompt}]
)

print("\n--- EXPANDED TEXT ---")
print(expand.choices[0].message.content)
```

5️⃣ Few-shot Example

```python
few_shot_prompt = f"""
Summarize the text in one sentence.

Example 1:
Text: AI is transforming industries by automating tasks.
Summary: AI is automating and transforming industries.

Example 2:
Text: Climate change is causing rising temperatures and extreme weather.
Summary: Climate change leads to higher temperatures and severe weather.

Now summarize:
Text: Percy Jackson receives a distress call from Grover and must rescue Artemis while facing a dangerous prophecy.
"""

few_shot = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": few_shot_prompt}]
)

print("\n--- FEW SHOT OUTPUT ---")
print(few_shot.choices[0].message.content)
```

Perform sentiment analysis on 5 customer reviews using BERT/Any Groq Model

Using BERT(Hugging Face)

```python
from transformers import pipeline

sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

reviews = [
    "The product quality is amazing and delivery was fast.",
    "Very disappointed with the service, not worth the money.",
    "Average experience, nothing special but okay.",
    "Excellent support team and great user experience!",
    "The product broke within two days, very poor quality."
]


print("---- BERT Sentiment Analysis ----")
for review in reviews:
    result = sentiment_pipeline(review)[0]
    print(f"\nReview: {review}")
    print(f"Sentiment: {result['label']} (Confidence: {result['score']:.2f})")
```

Using Groq Model (LLM-Based)

```python
from groq import Groq

client = Groq(api_key=groq_api_key)

reviews = [
    "The product quality is amazing and delivery was fast.",
    "Very disappointed with the service, not worth the money.",
    "Average experience, nothing special but okay.",
    "Excellent support team and great user experience!",
    "The product broke within two days, very poor quality."
]

print("---- Groq Sentiment Analysis ----")

for review in reviews:
    prompt = f"Classify the sentiment of this review as Positive, Negative, or Neutral:\n\n{review}"

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role": "user", "content": prompt}]
    )

    print(f"\nReview: {review}")
    print("Sentiment:", response.choices[0].message.content.strip())
```

Generate a product description using Llama.

```python
from groq import Groq

client = Groq(api_key=groq_api_key)


product_name = "Smart Fitness Watch"
features = "heart rate monitor, sleep tracking, waterproof, 7-day battery life, Bluetooth connectivity"

prompt = f"""
Generate a professional and attractive product description.

Product Name: {product_name}
Features: {features}

Make it engaging and suitable for an e-commerce website.
"""

response = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": prompt}],
    temperature=0.7,
    max_tokens=200
)

print("---- PRODUCT DESCRIPTION ----")
print(response.choices[0].message.content)
```

Summarize a news article using GPT-OSS.

```python
from groq import Groq

client = Groq(api_key=groq_api_key)

article = """
India successfully launched a new satellite aimed at improving weather forecasting and disaster management.
The satellite will provide real-time data to help authorities track storms, floods, and climate changes.
Officials say this advancement will significantly enhance early warning systems and reduce damage caused by natural disasters.
The mission is part of a broader initiative to strengthen space-based infrastructure and support sustainable development.
"""

prompt = f"""
Summarize the following news article in 4 concise bullet points:

{article}
"""

response = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": prompt}],
    temperature=0.4,
    max_tokens=150
)


print("---- NEWS SUMMARY ----")
print(response.choices[0].message.content)
```

Try zero-shot vs few-shot prompting and compare outputs

Zero-Shot Prompting

```python
from groq import Groq

client = Groq(api_key=groq_api_key)

review = "The product quality is decent but delivery was very slow."

zero_shot_prompt = f"""
Classify the sentiment of the following review as Positive, Negative, or Neutral:

Review: {review}
"""

response_zero = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": zero_shot_prompt}],
    temperature=0.3,
    max_tokens=50
)

print("---- ZERO SHOT OUTPUT ----")
print(response_zero.choices[0].message.content)
```

Few-Shot Prompting


```python
few_shot_prompt = f"""
Classify the sentiment as Positive, Negative, or Neutral.

Example 1:
Review: "Amazing product and fast delivery"
Sentiment: Positive

Example 2:
Review: "Very bad quality and poor service"
Sentiment: Negative

Example 3:
Review: "Product is okay but delivery is late"
Sentiment: Neutral

Now classify:
Review: {review}
"""

response_few = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[{"role": "user", "content": few_shot_prompt}],
    temperature=0.3,
    max_tokens=50
)

print("\n---- FEW SHOT OUTPUT ----")
print(response_few.choices[0].message.content)
```
